```python
run_sql("""
SELECT 
    t1.ticker,

    t1.volume,
    t2.volume,

    (t1.volume - t2.volume) AS volume_delta,

    (t1.volume - t2.volume) / t1.average_volume_30_days AS rel_volume_delta,

    t1.open,
    t1.high,
    t1.low,

    ((t1.high - t1.low) / t1.open) AS range_pct,

    t1.market_cap

FROM robinhood_fundamentals t1

JOIN robinhood_fundamentals t2
    ON t1.ticker = t2.ticker

WHERE 
    t1.scraped_at = (SELECT MAX(scraped_at) FROM robinhood_fundamentals)
    AND t2.scraped_at = (
        SELECT MAX(scraped_at)
        FROM robinhood_fundamentals
        WHERE scraped_at < (SELECT MAX(scraped_at) FROM robinhood_fundamentals)
    )

ORDER BY rel_volume_delta DESC
LIMIT 100
""")
```
