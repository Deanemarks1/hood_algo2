```python
# ============================================================
# PULL GITHUB FILES NOTEBOOK
# ============================================================

import os
import time
import requests
import hashlib

print("\n📡 GitHub Pull Utility Starting...\n")

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

repo_owner = "Deanemarks1"
repo_name = "sql_cookbook"
branch = "main"

files_to_pull = [
    "mysqlconnector.py",
    "selenium_setup.py",
    "deane_text_functions.py",
    "git_cookbook.ipynb"
]

auto_execute_py_files = True


# ------------------------------------------------------------
# BASE URL
# ------------------------------------------------------------

base_url = f"https://raw.githubusercontent.com/{repo_owner}/{repo_name}/{branch}/"


# ------------------------------------------------------------
# HELPER: FILE HASH (to detect changes)
# ------------------------------------------------------------

def file_hash(content_bytes):
    return hashlib.md5(content_bytes).hexdigest()


# ------------------------------------------------------------
# PULL LOOP
# ------------------------------------------------------------

for file_name in files_to_pull:

    print(f"📥 Pulling: {file_name}")

    # Cache bust
    url = base_url + file_name + f"?v={int(time.time())}"

    r = requests.get(
        url,
        headers={"Cache-Control": "no-cache"}
    )

    if r.status_code != 200:
        print(f"❌ Failed to pull {file_name}")
        continue

    remote_content = r.content
    remote_hash = file_hash(remote_content)

    # Check if local file exists
    if os.path.exists(file_name):
        with open(file_name, "rb") as f:
            local_hash = file_hash(f.read())
    else:
        local_hash = None

    # Compare
    if local_hash == remote_hash:
        print(f"🟡 No changes: {file_name}\n")
        continue

    # Overwrite
    with open(file_name, "wb") as f:
        f.write(remote_content)

    print(f"✅ Updated: {file_name}")

    # Auto execute .py files
    if auto_execute_py_files and file_name.endswith(".py"):
        with open(file_name) as f:
            code = compile(f.read(), file_name, "exec")
            exec(code, globals())
        print(f"⚡ Executed: {file_name}")

    print()



print("🔥 GitHub pull complete.\n")
```
