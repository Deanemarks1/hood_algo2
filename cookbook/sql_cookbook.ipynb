```python



#db Cookbook setup OCT 23
#-----------------------------------------------------------------------------------------------------------
import requests

# list of files to pull from your repo
files_to_pull = [
    "mysqlconnector.py",
    "git_cookbook.ipynb", 
    'selenium_setup.py', 
    'deane_text_functions.py'
    
]






base_url = "https://raw.githubusercontent.com/Deanemarks1/sql_cookbook/main/"

for file_name in files_to_pull:
    url = base_url + file_name
    print(f"📥 Pulling {file_name} from GitHub...")
    
    r = requests.get(url)
    r.raise_for_status()

    with open(file_name, "wb") as f:
        f.write(r.content)
        print(f"✅ Overwrote local copy: {file_name}")

    # execute only .py files automatically
    if file_name.endswith(".py"):
        with open(file_name) as f:
            code = compile(f.read(), file_name, "exec")
            exec(code, globals())
            print(f"⚡ Executed: {file_name}")

print("🔥 All GitHub files updated.")
#-----------------------------------------------------------------------------------------------------------





#Web app set up
#---------------------------------------------------------------------------------------##
#database = 'boo_by_boat'


#Select a Database
#if 'deanemarks' in base_dir:
     #run_sql(f'use {database}')

#if 'comradmarx' in base_dir:
    #run_sql(f'use comradmarx${database}')
#---------------------------------------------------------------------------------------##


#run_sql("USE scan2give_v2;")




```
